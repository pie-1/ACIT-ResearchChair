Matplotlib is a python library for creating charts and graphs. You can take plain numbers and turn them into visuals we can actually understand

With just a couple of lines of codes, we can make our first graph. We can create line charts, bar charts, scatter plots, histograms and pie charts.

We can customize the appearance of what our charts look like too.

```python
import matplotlib.pyplot as plt
```

pyplot is a module found inside library of matplotlib.

pyplot provides a user-friendly interface for plotting.

```python
print(matplotlib.__version__)  #__ or double underscore is called as dunder.
```

To create a very basic plot. We will need a set of x and y coordinates.

We will simply say x equals to a python list.

```python
x = [2023, 2024, 2025, 2026]
y = [15, 25, 30, 20]            #for each x year we have y no. of students.
    #Now to create a basic plot, we're going to accesss plt for plot() function and pass in our coordinates.
plt.plot(x, y)              #If we just run this nothing displays in regular python file, we also need show() function.
#       But since we are in jupyter notebook, it displays graaph without show() function.
plt.show()                   
```

Here is our graph. But we'll have to do some customization in future.

if we just pass in one argument in plt.plot() function, The default will automatically be assigned and it will start from 0 to 1, 2, 3 (it will also have 0.5 increment each) and soon for a missing argument. 

Now we can also pass in numpy arrays instead of lists because numpy arrays are way faster then regular python lists.

```python
import numpy as np
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.plot(y)             # one element in plot() to prove the previous point
plt.show()
```
