Bar charts: we know what they are:

It is a plot which compares categories of data by representing each category with a bar.

Lets create a bar chart to represent daily consumption of categories of foods.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = np.array(["protein", "sweets", "fruits", "vegetables", "grains", "dairy"])
# for every category of food what value should each of them have.
# These categories are for daily servings. How many in each category did I consume today. 
values = np.array([4, 3, 2, 5, 3, 1])
```

Categories will be in x-axis

Values will be on the y-axis

we use bar() function in which we pass x, y arguments.

```python
plt.bar(categories, values,
        color = "pink") #We can set colour of our choice, can be color's hex code as well.

#Lets add a title and two labels in our bar chart.
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

Note: 

    plt.title("Daily Consumption")
    plt.xlabel("Food")
    plt.ylabel("Quantity")

Must be in between plt.bar() and plt.show(), otherwise we will be applying these changes in another bar graph, instead of the one we need.

For a horizontal bar chart:

```python
plt.barh(categories, values, color ="pink")
plt.show()
```
