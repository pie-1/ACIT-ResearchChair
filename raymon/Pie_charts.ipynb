```python
import matplotlib.pyplot as plt
import numpy as np
```

Today we will be learning about pie charts. Pie charts are a circular charts that are divided into slices to show percentages of total. Good for visualizing distribution among categories.

create a fictional college. it must have a four classes: freshman, sophomore, junior. assign random students in each categories. display a pie chart.


```python
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])
colors = ["red", "yellow", "blue", "pink"]
```

when we call the pi function, we will pass in a few keyword arguments. 

They are values, labels(what are the labels in our pie chart).

We can use some percentages to each of the pie slices. The keyword argument is "autopct" which means auto percentage.

    autopct = "%1f" #autopct is set equal to a formatting instruction in which we use %1f.

    using %1 will result in greater precision so many digits after decimal. so its set to 1.1f

    %% to display percent sign because % is a format  specifier.


We can also explode a pie slice, explode means to highlight. Like with colors, it also has a keyword argument "explode". Each category is assigned a explosive number (I made this up for better understanding), greater the number, more distance the individual pie chart will be. For learning purpose, I exploded senior.

We can also assign drop shadow to pie slice by using "shadow= True"

We can also rotate the pie chart by using startangle= "Required degree of rotation" (To make a mirror view 180)

```python

plt.pie(values, labels = categories,
                autopct = "%1.1f%%",
                colors= colors,
                explode=[0, 0, 0, 0.2],
                shadow = True,
                startangle=180)

#Lets add a title to our pie chart;
plt.title("Acme Engineering college")

plt.show()
```
