Here we will be learning about customizing the labels, our title, and the ticks of our chart.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array(["1st sem", "2nd sem", "3rd Sem", "4th sem"])
y1 =np.array([9, 7.5, 4, 2])
y2 =np.array([11, 14, 14, 15])
y3 =np.array([14, 15, 14, 16])
y4 =np.array([15, 13, 15, 15])
plt.title("Each semester performance", fontsize= 15,
                                        family='Arial',
                                        fontweight="bold",
                                        color='black')  #We can use a HEXvalue for specific color.
plt.xlabel("Semester", fontsize = 10,
                    family='Arial',
                    fontweight="bold",
                    color='black')  
plt.ylabel("Performance", fontsize = 10,
                    family='Arial',
                    fontweight="bold",
                    color='black')                                        
plt.plot(x, y1, label="raymon")
plt.plot(x, y2, label="amrit")
plt.plot(x, y3, label="khushi")
plt.plot(x, y4, label="bhumi")
plt.legend(loc="center right")      #label="Student 1" assigns a name to the corresponding line.
#                                    plt.legend() displays those names in a legend.


#   Customizing the ticks of x and y axis
plt.tick_params(axis="y",
                colors='black')


plt.show()
```

Now we are going to learn about grid lines.

    grid() = Helps make the plot easier to read by adding reference lines

```python
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]

plt.grid(axis="both",           #axis can be x or y or both
        linewidth=2,
        color="lightgray",
        linestyle="dashed"
        )

plt.plot(x, y)
plt.show()
```
