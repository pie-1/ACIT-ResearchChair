```python
import matplotlib.pyplot as plt
import numpy as np
```

Scatter graph = Shows the relationship between two variables

    Helps to identify a correlation (+, -, None)

    Example: Study hours vs. Test scores

Lets take a example of a study hours and test scores of list of students.

    To create a scatter graph we are going to use the "scatter()" function.

    We use plt.show() to show the scatter graph.

```python
x = [0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8] # Hours studied
y = [55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87] # Grades


plt.scatter(x, y, color="Red", # From the 3rd argument, we begin customization.
                  alpha= 0.5,  # Alpha is transparency of dots, less alpha means less visible.
                  s = 100,     # s means size, we are changing size of dots.
                  label = "class A")
#   Lets add another dataset x2, y2 of hours studied and test scores.
x2 = [0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8] # Hours studied
y2 = [50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97] # Grades


plt.scatter(x2, y2, color="Blue", 
                  alpha= 0.5,  
                  s = 100,
                  label = "class B")     
                  


#   Lets add some labels to our scattergraph.
plt.title("Test Scores")
plt.xlabel("Hours studied") #For the x label, access plt, call the xlabel() and pass in string to serve as label.
plt.ylabel("Grade")

#   Note without accessing legend(), we can't show the label class A and class B for scatter graphs which we
#   have picked as 6th argument in scatter().
plt.legend()

plt.show()
```
