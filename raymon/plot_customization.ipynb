Here we will be learning how to customize our plots, more specifically, we'll be customizing the lines and markers. Here's how.

```python
import matplotlib.pyplot as plt
import numpy as np
```

We can use ., *, +, o, <, ^, s for square, p for pentagon, h for hexagon and soon for marker. 

We can set marker size as well, and it can be abbriviated to ms only. but for learning purposes, we use markersize. Same is true for markerfacecolor which can be written as mfc. mfc helps to pick a suitable colour for marker in line.

For line we can do following:

    linestyle can be dashed, dotted, none or solid (by default).
    
    We can also change linewidth.

    We can choose different color of line by simply passing color = "black" in plot()

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
plt.plot(x, y1, marker=".",
            markersize=25,
            markerfacecolor="black", 
            markeredgecolor="black",
            linestyle="solid",
            linewidth=4,
            color="black"
            )          
plt.plot(x, y2)
plt.show()
```

Here when we plot the coordinates (x, y1), (x, y2) both lines have different customizations.

One option to make them similar is by copy pasting the customization values from plot(x, y1) to plot(x, y2). This is Not recommended.

We can instead make a dictionary for our set of customization for lines and marker. 

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([15, 19, 7, 40])
line_style = dict(marker=".",          #We simply used 'dict()' constructor to create a dictionary in easier terms.
                  markersize=25,
                  markerfacecolor="black", 
                  markeredgecolor="black",
                  linestyle="solid",
                  linewidth=4,
                  color="black")
                                     #Now we can pass in this dictionary in plot()
plt.plot(x, y1, **line_style)          #But we do have to unpack the dictionary by **
plt.plot(x, y2, **line_style)
plt.plot(x, y3, **line_style)
plt.show()
```

Here all the lines have same color, but we can pass in individual distinct color by removing attribute of color from the line_style dictionary and passing it seperately in each plot() functions.

    Do ensure to remove color from dictionary as if we pass in individually when color attribute alr exists in dictionary, it will throw in error.


```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([15, 19, 7, 40])
line_style = dict(marker=".",          
                  markersize=15,
                  markerfacecolor="black", 
                  markeredgecolor="black",
                  linestyle="solid",
                  linewidth=4,
                  )
                                    
plt.plot(x, y1, **line_style, color = "green")      
plt.plot(x, y2, **line_style, color = "red")
plt.plot(x, y3, **line_style, color ="blue")
plt.show()
```
